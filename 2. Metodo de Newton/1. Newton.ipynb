## Función del Método de Newton

- f = función
- df = derivada
- x0 = 
- tol = tolerancia
- maxiter = cantidad maxima de iteraciones

```python
def Newton(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1
    
    while error>tol and iters<=maxiter:
        x = x0-f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters += 1
        x0 = x

    return [x, iters, error]
```

## Ejemplo 1

Resolver $x^3-x-1=0$ usando el método de newton con una estimación inicial de $x_0=-2$ y una tolerancia de $10^{-6}$

```python
#Definimos f

f = lambda x: x**3-x-1

# obtenermos la derivada
df = lambda x: 3*x**2-1

x0 = -2
tol = 1e-6
maxiter = 30
```

```python
x, iters, error = Newton(f,df,x0,tol,maxiter)

print(f"La solución obtenida fue {x}")
print(f"La cantidad de iteracione realizadas fueron {iters}")
print(f"El error que se obtuvo fue de {error}")
```
