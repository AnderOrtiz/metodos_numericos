```python
import math
import pandas as pd
```

## Función del método de Newton

```python
def Newton(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1

    while error>tol and iters<=maxiter:
        x = x0-f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        x0 = x
    return [x, iters, error]
```

## Función que brinda la tabla con todas las iteraciones

```python
def NewtonTabla(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1
    lista_e = list(); lista_x = list()
    
    while error>tol and iters<=maxiter:
        x = x0-f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        lista_e.append(error)
        lista_x.append(x)
        x0 = x
    Tabla = pd.DataFrame({"Iteraciones": lista_x, "Errores": lista_e})
    pd.set_option('display.precision', 10)
    return Tabla
```

## Ejemplo 

Usar el método de Newton para obtener soluciones con una tolerancia de $10^{-6}$ para la ecuación 
$\ln (x-1)+\cos (x-1)=0$ usando una estimación inicial $x_0=1.5$

```python
#Introducir f
f = lambda x: math.log(x-1)+math.cos(x-1)

#Indroducir df = 1/(x-1)-sen(x-1)
df = lambda x: 1/(x-1)-math.sin(x-1)

x0 = 1.5
tol = 1e-6
maxiter = 30

#Obtenemos los resultados
x,iters,error = Newton(f,df,x0,tol,maxiter)
```

```python
print("La solución a la ecuación es x = %s" %x)
print("El número de iteraciones fue de %s" %iters)
print("El error que se tuvo fue %s" %error)
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```

## Ejemplo 

Resolver la ecuación $$e^x+2^{-x}+2 \cos x-6=0$$
usando el método de Newton con $x_0=1.5$ y una tolerancia de $10^{-5}$

```python
#Introducir f
f = lambda x: math.exp(x)+2**(-x)+2*math.cos(x)-6

#Introducir la derivada df = e^x-2^(-x)ln(2)-2sin(x)
df = lambda x: math.exp(x)-2**(-x)*math.log(2)-2*math.sin(x)

x0 = 1.5 #Estimación inicial
tol = 1e-5 #Tolerancia
maxiter = 30

x,iters,error = Newton(f,df,x0,tol,maxiter)
```

```python
print("La solución a la ecuación es x = %s" %x)
print("El número de iteraciones fue de %s" %iters)
print("El error que se tuvo fue %s" %error)
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```

## Ejemplo

La suma de dos números es $20$. Si a cada uno se le agrega a su raíz cuadrada el producto de las dos sumas es 155.55. Determinar los dos números con un tolerancia de $10^{-6}$. 

```python
#Introducir f

f = lambda x: (x+math.sqrt(x))*(20-x+math.sqrt(20-x))-155.55

#Introducir df
df = lambda x: (x+math.sqrt(x))*(-1-1/2*(20-x)**(-1/2))+(20-x+math.sqrt(20-x))*(1-1/2*x**(-1/2))

#Estimación inicial
x0 = 10
tol = 1e-6
maxiter = 30

x,iters,error = Newton(f,df,x0,tol,maxiter)
```

```python
print("La solución a la ecuación es x = %s" %x)
print("El número de iteraciones fue de %s" %iters)
print("El error que se tuvo fue %s" %error)
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```

## Ejemplo

Al dejar caer un objeto con una masa $m$ desde una altura $s_0$ la altura se encuentra con la fórmula
$$s(t)=s_0-\dfrac{mg}{k}t+\dfrac{m^2g}{k^2}(1-e^{-kt/m})$$
donde $g=32.17$, $s_0=300$, $m=0.25$ y $k=0.1$. Obtener con una tolerancia de $10^{-5}$ el tiempo que tarda el objeto en tocar el suelo.

```python
f = lambda t: 300-80.425*t+201.0625*(1-math.exp(-0.4*t))

df = lambda t: -80.425+80.425*math.exp(-0.4*t)

x0 = 5
tol = 1e-5
maxiter = 30

x,iters,error = Newton(f,df,x0,tol,maxiter)

print(f"La solución a la ecuación es x = {x}")
print("El número de iteraciones fue de %s" %iters)
print("El error que se tuvo fue %s" %error)
```
