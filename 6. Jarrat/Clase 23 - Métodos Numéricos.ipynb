```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Función que implementa el método de Jarrat

```python
def Jarrat(f,df,x0,tol,maxiter):
    
    iters = 0
    error = tol+1
    error_x = list()

    while error>tol and iters<=maxiter:

        y = x0-2/3*f(x0)/df(x0)
        x = x0-(3*df(y)+df(x0))/(6*df(y)-2*df(x0))*f(x0)/df(x0)
        error = math.fabs(x-x0)
        error_x.append(error)
        iters = iters+1
        x0 = x

    error_x = np.array(error_x)
    ACOC = np.log(error_x[2:]/error_x[1:-1])/np.log(error_x[1:-1]/error_x[0:-2])
    
    return [x,iters,error,ACOC]
```

## Ejemplo 1:

Resolver la ecuación $x^2+\sin(x)+x=0$ usando el método de Jarrat, con una estimación inicial de $x_0=2$ y un tolerancia de $10^{-8}$.

```python
#Introducir f
f = lambda x: x**2+math.sin(x)+x

#Introducir la derivada
df = lambda x: 2*x+math.cos(x)+1
x0 = 2
tol = 1e-8
maxiter = 30

x,iters,error,ACOC = Jarrat(f,df,x0,tol,maxiter)

print("El valor de la solución es de x = %s" %x)
print("El número de iteraciones realizadas es de %s" %iters)
print("El error cometido fue %s" %error)
print("Los resultados del ACOC son %s" %ACOC)
```

## Función que muestra los resultados en forma de tabla

```python
def JarratTabla(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1
    lista_e = list(); lista_x = list()
    
    while error>tol and iters<=maxiter:
        y = x0-2/3*f(x0)/df(x0)
        x = x0-(3*df(y)+df(x0))/(6*df(y)-2*df(x0))*f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        lista_e.append(error)
        lista_x.append(x)
        x0 = x
    Tabla = pd.DataFrame({"Iteraciones": lista_x, "Errores": lista_e})
    pd.set_option('display.precision', 16)
    return Tabla
```

JarratTabla(f,df,x0,tol,maxiter)

## Función que implementa el método de CMT

```python
import mpmath as mp

mp.mp.dps = 100

def CMT(f,df,x0,tol,maxiter):
    
    iters = 0
    error = tol+1
    error_x = list()

    while error>tol and iters<=maxiter:

        y = x0-f(x0)/df(x0) #Newton
        z = y-f(x0)/(f(x0)-2*f(y))*f(y)/df(x0) #Ostrowki
        x = z-(1+2*f(y)/f(x0))*f(z)/df(x0)
        error = math.fabs(x-x0)
        error_x.append(error)
        iters = iters+1
        x0 = x

    error_x = np.array(error_x)
    ACOC = np.log(error_x[2:]/error_x[1:-1])/np.log(error_x[1:-1]/error_x[0:-2])
    
    return [x,iters,error,ACOC]
```

## Función que muestra los resultados en forma de tabla

```python
def CMTTabla(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1
    lista_e = list(); lista_x = list()
    
    while error>tol and iters<=maxiter:
        y = x0-f(x0)/df(x0)
        z = y-f(x0)/(f(x0)-2*f(y))*f(y)/df(x0)
        x = z-(1+2*f(y)/f(x0))*f(z)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        lista_e.append(error)
        lista_x.append(x)
        x0 = x
    Tabla = pd.DataFrame({"Iteraciones": lista_x, "Errores": lista_e})
    pd.set_option('display.precision', 16)
    return Tabla
```

## Ejemplo 4

Encontrar la solución estimada a la ecuación $(x+2)e^{x}-1=$ empleando el método CMT con una tolerancia de $10^{-10}$ y una estimación inicial $x_0=-1$

```python
#Introducir f
f = lambda x: (x+2)*math.exp(x)-1

#Introducir la derivada
df = lambda x: (x+2)*math.exp(x)+ math.exp(x)
x0 = mp.mpf('-1')
#x0 = -1
tol = mp.mpf('1e-10')
#tol = 1e-10
maxiter = 30

x,iters,error,ACOC = CMT(f,df,x0,tol,maxiter)

print("El valor de la solución es de x = %s" %round(x,16))
print("El número de iteraciones realizadas es de %s" %iters)
print("El error cometido fue %s" %error)
print("Los resultados del ACOC son %s" %ACOC)
```

```python
CMTTabla(f,df,x0,tol,maxiter)
```

## Ejemplo 5
