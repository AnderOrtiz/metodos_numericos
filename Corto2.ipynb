```python
import math
import numpy as np
import pandas as pd
```

```python
def Biseccion(f,a,b,tol,maxiter):
    iters = 0
    error = tol+1

    while error>tol and iters<=maxiter:
        x = (a+b)/2
        error = (b-a)/2
        if f(a)*f(x)>0:
            a = x
        else:
            b = x
        iters = iters+1

    return [x,iters,error]
```

```python
def Bisecciontabla(f,a,b,tol,maxiter):
    iters = 0
    error = tol+1
    lista_a = list(); lista_b = list()
    lista_a.append(a); lista_b.append(b);
    lista_x = list(); lista_e = list()

    while error>tol and iters<=maxiter:
        x = (a+b)/2
        error = (b-a)/2
        lista_e.append(error); lista_x.append(x)
        if f(a)*f(x)>0:
            a = x
        else:
            b = x
        iters = iters+1
        lista_a.append(a); lista_b.append(b);
    lista_a.pop(); lista_b.pop()
    tabla = pd.DataFrame({'a': lista_a, 'b': lista_b, 'x': lista_x, 'Errores': lista_e})
    pd.set_option('display.precision', 2)
    return tabla
```

### Definición de $f(x)$ y su derivada
---
$$ f(x) = x^2 - 100x\sin(x)+(10\sin(x))^2$$
---
$$f'(x) = 2x - 100\sin(x) - 100x\cos(x) + 200\sin(x)\cos(x)$$

```python
f = lambda x: x**2 - 100*x*math.sin(x) + (10*math.sin(x))**2
df = lambda x: 2*x - 100*math.sin(x) - 100*x*math.cos(x) + 200*math.sin(x)*math.cos(x)
```

### $f(x) = 0$ en $[2.1, 3.2]$

```python
Bisecciontabla(f, 2.1, 3.2, 1e-3, 30)
```

### $f'(x) = 0$ en $[-3, -2]$

```python
Bisecciontabla(df, -3, -2, 1e-3, 30)
```
