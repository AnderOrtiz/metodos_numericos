```python
def Biseccion(f,a,b,tol,maxiter):
    iter = 0
    error = tol + 1

    while iter < maxiter and error > tol:
        x = (a + b)/2
        error = (b - a) /2
        
        if f(a)*f(x)>0:
            a = x
        else:
            b = x

        iter += 1
    return [x, iter, error]
```

- f(x)=x^3 + 4x^2 - 10
- a=1, b=2
- tol= 1e^-3

```python
f = lambda x: x**3 + 4*x**2 - 10
a = 1
b = 2
tol = 1e-3 #10^(-3))
maxiter = 30
```

```python
[x, iter, error] = Biseccion(f,a,b,tol,maxiter)

print(f"Valor medio: {x}")
print(f"Número de iteraciones {iter}")
print(f"Error: {error}")
```
