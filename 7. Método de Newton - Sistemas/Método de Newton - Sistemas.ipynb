# Método de Newton para sistemas de ecuaciones no lineales

```python
import math
import numpy as np
import pandas as pd
```

## Función que implementa el método de Newton

```python
def NewtonSis(F,JF,x0,tol,maxiter):
    iters = 0
    error = tol + 1
    
    while iters<=maxiter and error > tol:
        z = np.linalg.solve(JF(x0), F(x0))
        x = x0 - z
        error = np.linalg.norm(x-x0)
        iters += 1
        x0 = x
    return [x,iters,error]
```

## Función que muestra la tabla del método de Newton

```python
def NewtonSisTabla(F,JF,x0,tol,maxiter):
    iters = 0 #Contador de iteraciones a cero
    error = tol+1
    lista_error = list()
    lista_x = list()
    while error>tol and iters<=maxiter:
        z = np.linalg.solve(JF(x0),F(x0)) #Se resuelve el sistema lineal
        x = x0-z
        lista_x.append(x)
        error = np.linalg.norm(x-x0) #Se encuentra el error con la norma
        lista_error.append(error)
        x0 = x
        iters = iters+1
    Tabla = pd.DataFrame()
    Tabla["x"] = [w[0] for w in lista_x]
    Tabla["y"] = [w[1] for w in lista_x]
    Tabla["Errores"] = lista_error
    pd.set_option('display.precision', 16)
    return Tabla
```

### Ejemplo 1

Encontrar la solución al sistemas de ecuaciones
\begin{align*}
x^2+xy-10&=0 \\
y+3xy^2-57&=0
\end{align*}

Con el método de Newton usando las estimaciones iniciales $x_0=1.5$ e $y_0=3.5$ y una tolerancia de $10^{-8}$

```python
def F(w):
    x,y = w
    S = np.zeros(2)
    S[0] = x**2+x*y-10
    S[1] = y+3*x*y**2-57
    # print(S.shape)
    return S

def JF(w):
    x, y = w
    M = np.zeros([2,2])
    M[0,0] = 2*x+y
    M[0,1] = x
    M[1,0] = 3*y**2
    M[1,1] = 1+6*x*y
    return M

tol = 1e-8
x0 = np.array([1.5,3.5])
maxiter = 30
```

```python
x, iters, error = NewtonSis(F,JF,x0,tol,maxiter)

print(f"La solición al sistema es {x}")
print(f"El error obtenido fue de {error}")
print(f"Las iteraciones realizadas fueron {iters}")
```

```python
NewtonSisTabla(F,JF,x0,tol,maxiter)
```

## Ejemplo 2

Resolver el siguiente sistema no lineal
\begin{align*}
4x^2-20x+\frac{1}{4}y^2+8 &=0 \\
\dfrac{1}{2}xy^2+2x-5y+8&=0
\end{align*}
Con una tolerancia de $10^{-6}$ y una estimación inicial de $\mathbf{x}^{(0)}=(0,0)$

```python
# Sistema no lineal 
def F(w):
    x,y = w
    S = np.zeros(2)
    S[0] = 4*x**2-20*x+1/4*y**2+8
    S[1] = 1/2*x*y**2+2*x-5*y+8
    return S

def JF(w):
    x,y = w
    M = np.zeros([2,2])
    M[0,0] = 8*x-20; M[0,1] = 1/2*y
    M[1,0] = 1/2*y**2; M[1,1] = x*y-5
    return M

tol = 1e-6
x0 = np.array([0,0])
maxiter = 30

x,iters, error = NewtonSis(F,JF,x0,tol,maxiter)

print(f"La solición al sistema es {x}")
print(f"El error obtenido fue de {error}")
print(f"Las iteraciones realizadas fueron {iters}")
```

```python
NewtonSisTabla(F,JF,x0,tol,maxiter)
```
