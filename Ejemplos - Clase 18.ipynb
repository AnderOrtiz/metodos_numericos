```python
import math
import pandas as pd
```

```python
def Newton(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1

    while error>tol and iters<=maxiter:
        x = x0-f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        x0 = x
    return [x, iters, error]
```

```python
def NewtonTabla(f,df,x0,tol,maxiter):
    iters = 0
    error = tol+1
    lista_e = list(); lista_x = list()
    
    while error>tol and iters<=maxiter:
        x = x0-f(x0)/df(x0)
        error = math.fabs(x-x0)
        iters = iters+1
        lista_e.append(error)
        lista_x.append(x)
        x0 = x
    Tabla = pd.DataFrame({"Iteraciones": lista_x, "Errores": lista_e})
    pd.set_option('display.precision', 10)
    return Tabla
```

## Ejemplo

Usar el método de Newton para resolver la ecuación $x^3-3x-5=0$ con una tolerancia de $10^{-6}$ con una estimación inicial de $x_0=2$

```python
#Introducir f
f = lambda x: x**3-3*x-5

#Introducir la derivada
df = lambda x: 3*x**2-3
x0 = 2
tol = 1e-6
maxiter = 30
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```

```python
#Introducir f
f = lambda x: math.log(x-1)+math.cos(x-1)
#Introducir la derivada
df = lambda x: 1/(x-1)-math.sin(x-1)
x0 = 1.5
tol = 1e-6
maxiter = 30

x, iters, error = Newton(f,df,x0,tol, maxiter)
```

```python
print(f"La solución a la ecuación es x = {x}")
print(f"El número de la iteraciones fue = {iters}")
print(f"El error que obtuvo fue = {error}")
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```

# Ejemplo

Resolver la ecuación

$e^x+2^{-x}+2cosx-6=0$

```python
# introducir f
f = lambda x: math.exp(x)+2**(-x)+2*math.cos(x)-6

# introducir la derivada = e^x-2(-x)ln(2)2sin(x)

df = lambda x: math.exp(x)-2**(-x)*math.log(2)-2*math.sin(x)
x0 = 1.5
tol = 1e-6
maxiter = 30

x, iters, error = Newton(f,df,x0,tol, maxiter)

print(f"La solución a la ecuación es x = {x}")
print(f"El número de la iteraciones fue = {iters}")
print(f"El error que obtuvo fue = {error}")
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```
